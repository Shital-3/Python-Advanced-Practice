```python
# Learning Pandas
# It help to read,store,clean and analyze as well as summarize data
#Main Data structure in pandas
# 1)DataFrame- 2D data(like table with rows and col)
import pandas as pd
import numpy as np 
```

```python
## working with DataFrame
df=pd.DataFrame(
    np.arange(0,12).reshape(4,3),
    index=['Row1', 'Row2', 'Row3', 'Row4'],
    columns=['col1', 'col2', 'col3']

)

print(df)
```

```python
# to_csv() is used to save your df into a csv file
# csv=comma separated values
# Basic syntax is df.to_csv("filename.csv)
df.to_csv("Test1.csv")
```

```python
#Accessing the elements from dataframe -> there are two ways to do this
# first way
df.loc['Row1'] # to acces row
```
