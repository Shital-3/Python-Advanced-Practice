```python
import numpy as np
import matplotlib.pyplot as plt
x=np.arange(0,10)
y=np.arange(11,21)
# plot scatter
plt.scatter(x,y,c='g') # scatter meand draw a dotted line 

#x= horizontal position
#y=vertical position
#c='g' means : color =green
plt.xlabel('X axis') #Write X axis below the horizontal line.
plt.ylabel('Y axis')#Write Y axis beside the vertical line.
plt.title('Graph in 2D') # add title to graph
#plt.savefig('Test.png') #Save the Figure as Image

```

```python
y=x*x

plt.plot(x,y,'r*',linestyle='dashed',linewidth=2, markersize=12)
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('2d Diagram')
```

```python
plt.subplot(2,2,1)
#2 rows
# columns
#draw in box number 1

plt.plot(x,y,'r--')
#'r--'->red dashed line
#'g*--'->green stars + dashed line
#'bo'->blue circle
# 'go'-> green circles

plt.subplot(2,2,2)
#You are always saying:

#Use the SAME 2×2 grid,
#just change which box I am drawing in.


plt.plot(x,y,'g*--')
plt.subplot(2,2,3)
plt.plot(x,y,'bo')

plt.subplot(2,2,4)
plt.plot(x,y,'go')
```

```python
#compute the x and y coordinate for points on the sine curve
x=np.arange(0,4*np.pi,0.1) #0, 0.1, 0.2, 0.3, ... up to about 12.56
# above is  a ruler starting from 0 and going up to 4π
y=np.sin(x)
plt.title("sine wave form")
plt.plot(x,y)
plt.show()
```

```python

# plot the Bar
x=[2,8,10]
y=[11,16,9]
#At position:
#x = 2 → height = 11
#x = 8 → height = 16
#x = 10 → height = 9
#so Matplotlib will draw 3 bars
x2 = [3,9,11] 
y2 = [6,15,7]
plt.bar(x, y)# default color= blue
plt.bar(x2, y2, color='g')
plt.title('Bar graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()


```

```python
 # Histograms
#histogram shows that how many numbers fall in each group,
#not exact values, but how many values are small,medium and big etc.
a=np.array([22,87,5,43,56,73, 55, 54, 11, 20, 51, 5, 79, 31, 27])
# it may be marks of student or age etc.
plt.hist(a)
plt.title("histogram")
plt.show()
#Taller bar = more numbers in that range
#Short bar = fewer numbers in that range
```

```python
# pie chart
labels='Python','C++','Ruby','Java'
sizes=[215,130,245,210]
#These numbers tell how big each slice should be.
colors = ['gold', 'yellowgreen', 'lightcoral', 'lightskyblue']
explode = (0.4, 0, 0, 0)
#first slice will be pulled out a little,other slices stay normal
# Draw the pie
plt.pie(
    sizes,
    explode=explode,
    colors=colors,
    autopct='%1.1f%%',
    shadow=False
)
```
